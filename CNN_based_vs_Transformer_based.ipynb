## 🔎 Background

Across both binary and multiclass classification tasks, several models demonstrated strong performance.  
To confirm whether these performance differences were statistically meaningful, we conducted **paired *t*-tests** on fold-level accuracies obtained from 5-fold cross-validation.  

The paired *t*-test is suitable here because it directly compares the same folds across two models, controlling for variability introduced by different train–test splits.


## 📂 1. Binary Classification (ViT vs. VGG16)

We compare:
- **ViT (Vision Transformer)** – Transformer-based  
- **VGG16** – CNN-based baseline  

Both models were trained on the binary classification task (Mpox vs. non-Mpox)using **5-fold CV**.


```python
import numpy as np
from scipy import stats

# Binary classification accuracies
vit_acc = np.array([0.8750, 0.8511, 0.7750, 0.8889, 0.8958])
vgg16_acc = np.array([0.8125, 0.7872, 0.7000, 0.8000, 0.7917])

# Paired t-test
t_stat, p_val = stats.ttest_rel(vit_acc, vgg16_acc)
print("ViT vs VGG16 (Binary) -> t=%.4f, p=%.4f" % (t_stat, p_val))
```

## 📂 2. Multiclass Classification (ResNetViT vs. ResNet50)

We compare:
- **ResNetViT (Transformer-based hybrid)**  
- **ResNet50 (CNN baseline)**  

Both trained and tested under **5-fold CV** for the multiclass task (6 classes).


```python
# Multiclass classification accuracies
resnetvit_acc = np.array([0.8472, 0.9024, 0.8301, 0.8600, 0.8247])
resnet50_acc = np.array([0.8125, 0.7805, 0.6928, 0.7600, 0.7468])

# Paired t-test
t_stat, p_val = stats.ttest_rel(resnetvit_acc, resnet50_acc)
print("ResNetViT vs ResNet50 (Multiclass) -> t=%.4f, p=%.4f" % (t_stat, p_val))
```

### 📈 Interpretation

- **Result:** ViT significantly outperforms VGG16 (*p < 0.05*).  
- This shows that transformer-based models deliver **statistically reliable gains** compared to CNN baselines in binary classification.

- **Result:** ResNetViT significantly outperforms ResNet50 (*p < 0.01*).  
- Confirms that transformer-based hybrids scale better to multiclass settings compared to CNN baselines.
